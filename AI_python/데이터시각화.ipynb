전처리의 중요성

 * 데이터가 정확할수록 분석과 집계 과정의 오류가 줄어듦
 * 잘못된 데이터를 입력하면 결과도 잘못될 수 있음
 * 데이터 전처리
   * 결측치 보정, 이상치 처리, 정규화, 인코딩을 수행하는 작업

데이터 전처리

 * 전처리
   * 원본 데이터를 분석과 모델 학습에 맞게 다듬는 작업
 * 데이터 불러오기, 전처리, 분석, 시각화, 모델 학습 순서로 진행
 * 입력 데이터의 품질은 분석과 모델 결과에 영향을 줌
 * 원본 데이터는 보존하고 전처리한 사본을 사용

전처리 4대 작업

 * 결측치(Missing Value)
   * 비어 있는 값을 채우거나 해당 데이터를 제거
 * 이상치(Outlier)
   * 다른 값보다 지나치게 크거나 작은 값을 확인하고 처리
 * 정규화(Normalization)
   * 서로 다른 값의 범위를 같은 기준으로 조정
 * 인코딩(Encoding)
   * 범주형 데이터를 숫자로 변환

결측치와 이상치 확인

 * df.isull().sum()
   * 컬럼별 결측치 개수를 확인
 * df.describe()
   * 최솟값과 최댓값 등으로 이상치의 단서를 확인
 * 상자그림과 히스토그램으로 값의 분포를 확인
 * 값을 처리하기 전에 결측치와 이상치가 발생한 원인을 확인
